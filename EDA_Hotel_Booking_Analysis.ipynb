```python
missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])
```

```python
print("Duplicate Records:", df.duplicated().sum())
```

```python
clean_df = df.drop_duplicates().copy()

print("Original Rows:", len(df))
print("Cleaned Rows:", len(clean_df))
```

```python
print(clean_df.isnull().sum())
```

```python
missing_percentage = (
    clean_df.isnull().sum() / len(clean_df)
) * 100

print(missing_percentage[missing_percentage > 0])
```

```python
clean_df["children"] = clean_df["children"].fillna(0)

print("Missing Children:", clean_df["children"].isnull().sum())
```

```python
clean_df["country"] = clean_df["country"].fillna("Unknown")

print("Missing Country:", clean_df["country"].isnull().sum())
```

```python
clean_df["agent"] = clean_df["agent"].fillna(0)
clean_df["company"] = clean_df["company"].fillna(0)

print("Missing Agent:", clean_df["agent"].isnull().sum())
print("Missing Company:", clean_df["company"].isnull().sum())
```

```python
clean_df["arrival_date"] = pd.to_datetime(
    clean_df["arrival_date_year"].astype(str) + "-" +
    clean_df["arrival_date_month"] + "-" +
    clean_df["arrival_date_day_of_month"].astype(str),
    format="%Y-%B-%d",
    errors="coerce"
)

clean_df[["arrival_date"]].head()
```

```python
clean_df["total_guests"] = (
    clean_df["adults"] +
    clean_df["children"] +
    clean_df["babies"]
)

print(clean_df["total_guests"].head())
```

```python
clean_df["total_nights"] = (
    clean_df["stays_in_weekend_nights"] +
    clean_df["stays_in_week_nights"]
)

print(clean_df["total_nights"].head())
```

```python
hotel_count = clean_df["hotel"].value_counts()

print(hotel_count)

hotel_count.plot(kind="bar")

plt.title("Bookings by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")
plt.show()
```

```python
cancellation_count = clean_df["is_canceled"].value_counts()

print(cancellation_count)

cancellation_count.plot(kind="bar")

plt.title("Booking Cancellation Analysis")
plt.xlabel("Cancellation Status (0 = No, 1 = Yes)")
plt.ylabel("Number of Bookings")
plt.show()
```

```python
customer_count = clean_df["customer_type"].value_counts()

print(customer_count)

customer_count.plot(kind="bar")

plt.title("Customer Type Distribution")
plt.xlabel("Customer Type")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=45)
plt.show()
```

```python
monthly_bookings = clean_df.groupby(
    "arrival_date_month"
).size()

print(monthly_bookings)

monthly_bookings.plot(kind="bar", figsize=(10, 5))

plt.title("Monthly Hotel Bookings")
plt.xlabel("Month")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=45)
plt.show()
```

```python
room_count = clean_df["reserved_room_type"].value_counts()

print(room_count)

room_count.plot(kind="bar", figsize=(8, 5))

plt.title("Reserved Room Type Distribution")
plt.xlabel("Room Type")
plt.ylabel("Number of Bookings")
plt.show()
```

```python
country_count = clean_df["country"].value_counts().head(10)

print(country_count)

country_count.plot(kind="bar", figsize=(10, 5))

plt.title("Top 10 Countries by Bookings")
plt.xlabel("Country")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=45)
plt.show()
```

```python

print("Average Daily Rate:", clean_df["adr"].mean())
print("Minimum Daily Rate:", clean_df["adr"].min())
print("Maximum Daily Rate:", clean_df["adr"].max())
```

```python
hotel_adr = clean_df.groupby("hotel")["adr"].mean()

print(hotel_adr)

hotel_adr.plot(kind="bar")

plt.title("Average Daily Rate by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Average Daily Rate")
plt.show()
```

```python
guest_count = clean_df["total_guests"].value_counts().sort_index()

print(guest_count.head(10))

guest_count.head(10).plot(kind="bar")

plt.title("Bookings by Number of Guests")
plt.xlabel("Total Guests")
plt.ylabel("Number of Bookings")
plt.show()
```

```python
print("HOTEL BOOKINGS EDA SUMMARY")
print("--------------------------")

print("Total Bookings:", len(clean_df))

print("City Hotel Bookings:",
      (clean_df["hotel"] == "City Hotel").sum())

print("Resort Hotel Bookings:",
      (clean_df["hotel"] == "Resort Hotel").sum())

print("Cancelled Bookings:",
      clean_df["is_canceled"].sum())

print("Average Daily Rate:",
      round(clean_df["adr"].mean(), 2))

print("Average Guests:",
      round(clean_df["total_guests"].mean(), 2))

print("Average Stay Nights:",
      round(clean_df["total_nights"].mean(), 2))

print("Most Common Customer Type:",
      clean_df["customer_type"].mode()[0])

print("Most Common Room Type:",
      clean_df["reserved_room_type"].mode()[0])

print("\nHOTEL EDA COMPLETED!")
```
